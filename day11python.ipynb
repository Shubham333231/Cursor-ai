```python
#write a function that  return the reversed number

def reversed_num(n):
  total = 0

  while n > 0 :
    digits = n % 10
    total =  total * 10 + digits
    n = n // 10

  return total

reversed_num(12345)
```

```python
#write a function checks palindrome

def palindrome_check(n):
  original = n
  total = 0

  while n > 0:
    digits = n % 10
    total = total * 10 + digits
    n = n // 10

  return total == original

palindrome_check(12321)
```

```python
#write a function that returns sum of all digits (digital sum)

def sum_of_digits(n):
  total = 0

  while n > 0:
    digit = n % 10
    total = total + digit
    n = n // 10

  return total

print(sum_of_digits(12345))


```

```python
#write a function count_even that counts how many even numbers exists from 1 to n


def count_even(n):
  count = 0
  for i in range(1,n+1):
    if i % 2 == 0:
      count += 1
  return count

print(count_even(10))

```

```python
#write a function that checks wheather a number is prime

def prime_check(n):
  if n < 2:
    return False

  i = 2
  while i < n:
    if n % i == 0:
      return False
    i = i + 1

  return True



```

```python
prime_check(9)
```

```python
#write a function that prints all numbers between two numbers

def my_prime(n):
  if n < 2:
    return False

  i = 2

  while i < n:
    if n % i == 0:
      return False
    i = i + 1

  return True


for number in range(1,101):
  if my_prime(number):
    print(number)
```

```python
#write a funtion that returns and prints the fibonancci number n

def fibonacci(n):
  if n <= 0:
    return "Input should be a positive integer"
  elif n == 1:
    return 0
  elif n == 2:
    return 1
  else:
    a, b = 0, 1
    for _ in range(2, n):
      a, b = b, a + b
    return b

# Demonstrate the function
print(f"The 1st Fibonacci number is: {fibonacci(1)}")
print(f"The 2nd Fibonacci number is: {fibonacci(2)}")
print(f"The 10th Fibonacci number is: {fibonacci(10)}")
```
