```python
# print all subjects and their marks

marks = {"Maths": 85, "Python":92, "Physics":78, "English":88}
for subject, mark in marks.items():
  print(subject, ":" , mark)
```

```python
#print the total marks

marks = {"Maths": 85, "Python":92, "Physics":78, "English":88}
total_marks = sum(marks.values())
print(total_marks)
```

```python
#print average marks

marks = {"Maths": 85, "Python":92, "Physics":78, "English":88}
total_marks = sum(marks.values())
average_marks = total_marks / len(marks)
print(average_marks)
```

```python
#find the subject with highest marks without using max()

marks = {"Maths": 85, "Python":92, "Physics":78, "English":88}
highest_marks = 0
highest_subject = ""
for subject, mark in marks.items():
  if mark > highest_marks:
    highest_marks = mark
    highest_subject = subject
print(highest_subject)
```
