```python
def lesser_of_two_events(a,b):
  if a % 2 == 0 and b % 2 == 0:
    return min(a,b)
  else:
      return max(a,b)
```

```python
lesser_of_two_events(2,4)
```

```python
lesser_of_two_events(2,7)
```

```python
def animal_crackers(text):
  wordlist = text.split()

  return  wordlist[0][0] == wordlist[1][0]
```

```python
animal_crackers("Lion Leopard")
```

```python
animal_crackers("Crazy Sum")
```

```python
def function_my(name):

  first_letter = name[0]
  in_between = name[1:3]
  fourth_letter = name[3]
  rest = name[4:]

  return first_letter.upper() + in_between + fourth_letter.upper() + rest
```

```python
 function_my("macdonald")
```

```python
def my_macdonald(name):
  return " ".join(name.split()[::-1])
```

```python
my_macdonald("We are home")
```

```python
def almost_there(n):
  return ((abs(100 - n) <= 10) or (abs(200 - n) <= 10))
```
