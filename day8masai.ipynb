```python
def my_factorial(a):
  result = 1

  while a > 0:
    result  = result * a
    a = a-1

  return result
```

```python
my_factorial(5)
```

```python
def my_prime(n):
  if n < 2:
    return False

  i = 2

  while i < n:
    if n % i == 0:
      return False
    i = i + 1

  return True

sum_prime = 0

for number in range(1,101):
  if my_prime(number):
    print(number)
    sum_prime = sum_prime + number

print(sum_prime)
```
