```python
def check_number(a):
  if a > 0:
    return "Positive"
  elif a < 0:
    return "Negative"
  else:
    return "Zero"

print(check_number(1))
```

```python
def max_number(a,b):
  if a > b:
    return "a"
  else:
    return "b"

print(max_number(2,1))
```

```python
def sum_number(n):
  total = 0

  while n>0:
    digit = n % 10
    total = total + digit
    n = n // 10

  return total

print(sum_number(1234))
```

```python
def prime_num(n):
  if n < 2:
      return False

  i = 2

  for n in range (2,n):
      if n % i == 0:
          return False
      i = i + 1


  return True
print(prime_num(17))
```

```python
def reversed_num(n):
  reverse = 0

  while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

  return reverse

print(reversed_num(12345))
```

```python
def factorial(n):
  result = 1

  while n > 0:
    result = result * n
    n = n - 1

  return result
```

```python
factorial(5)
```

```python
def calc_sum(n):
  total = 0

  for n in range(1,n+1):
    total += 1

  return total
```

```python
 calc_sum(4)
```
