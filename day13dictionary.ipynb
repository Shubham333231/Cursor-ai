```python
def prime_check(n):
  if n < 2:
    return False

  i = 2
  while i > n:
    if n % i == 0:
      return True
    i = i + 1

  return True
```

```python
prime_check(9)
```

```python
#write a function that finds the largest digit

def largest_dig(n):
  largest = 0

  while n > 0:
    digit = n % 10

    if digit > largest:
      largest = digit

    n = n // 10

  return largest

print(largest_dig(12345))
```

```python
#write a function that accepts a dictionary of marks and returns the sum of all values.

marks = {"Maths": 85, "Python":92, "Physics":78, "English":88}

def sum_values(data):
  total = 0

  for value in data.values():
    total = total + value

  return total

print(sum_values(marks))

```

```python
#write a function that counts how many values are even

def count_even(n):
  count = 0

  for value in n.values():
    if value % 2 == 0:
      count = count + 1
  return count

marks = {"Maths": 85, "Python":92, "Physics":78, "English":88}

print(count_even(marks))
```

```python
#create a function that creates a new dictionary containing only subjects whose marks are greater than 80

def greater_than80(n):
  newcount  = 0
  newdict = {}

  for key, value in n.items():
    if value > 80:
      newdict[key] = value

  return newdict

marks = {"Maths": 85, "Python":92, "Physics":78, "English":88}

print(greater_than80(marks))
```
